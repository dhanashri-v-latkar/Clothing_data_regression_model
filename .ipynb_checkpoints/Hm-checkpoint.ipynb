```python
def data_cleaning(ser):

    # lowercase
    ser = ser.str.lower()

    # remove digits and punctuation
    ser = ser.str.replace(r'[^a-zA-Z\s]', ' ', regex=True)

    # remove extra spaces
    ser = ser.str.replace(r'\s+', ' ', regex=True).str.strip()

    # keep important sentiment words
    stop_words = nlp.Defaults.stop_words - {
        "not", "no", "never"
    }

    # lemmatization + stopword removal
    ser = ser.apply(
        lambda x: ' '.join(
            token.lemma_
            for token in nlp(x)
            if token.text not in stop_words
            and not token.is_punct
            and len(token.text) > 2
        )
    )

    return ser
```
